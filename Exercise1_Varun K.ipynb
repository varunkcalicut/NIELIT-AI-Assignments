# Exercise 1: Exploratory Data Analysis & Automated Reporting with Pandas
**Goal:** create a DataFrame with random data, plot f2 vs f3, compute the row-wise mean as `f5`, and export everything to Excel.

## Step 1: Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 2: Create a DataFrame with four columns of random numbers
`np.random.rand(50, 4)` gives 50 rows x 4 columns of values between 0 and 1. The seed makes the output repeatable.

```python
np.random.seed(42)
df = pd.DataFrame(np.random.rand(50, 4), columns=["f1", "f2", "f3", "f4"])
df.head()
```

## Step 3: Scatter plot of f2 vs f3

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["f2"], df["f3"], color="steelblue", alpha=0.7)
plt.xlabel("f2")
plt.ylabel("f3")
plt.title("Scatter Plot: f2 vs f3")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

## Step 4: Row-wise mean stored in `f5`
`axis=1` tells pandas to average across columns (one mean per row).

```python
df["f5"] = df[["f1", "f2", "f3", "f4"]].mean(axis=1)
df.head()
```

## Step 5: Export to Excel
Needs the `openpyxl` package (`pip install openpyxl`). `index=False` keeps the row index out of the sheet.

```python
df.to_excel("Exercise1_output.xlsx", index=False)

# Verify: read the file back
pd.read_excel("Exercise1_output.xlsx").head()
```

## Observations
- The DataFrame has 50 rows and 5 columns (f1 to f5).
- The scatter plot shows no clear pattern between f2 and f3, as expected for independent random values.
- Each f5 value equals the average of f1, f2, f3 and f4 in that row, and all five columns are saved in `Exercise1_output.xlsx`.
