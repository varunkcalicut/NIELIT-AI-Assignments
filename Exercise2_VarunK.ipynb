# Exercise 2 - Image Blending using OpenCV

## 1. Importing libraries

```python
import cv2
import matplotlib.pyplot as plt
```

## 2. Loading the two images
image1.jpg and image2.jpg are kept in the same folder as this notebook.

```python
img1 = cv2.imread("image1.jpg")
img2 = cv2.imread("image2.jpg")

# imread gives None if the file is not found, so checking this
if img1 is None or img2 is None:
    print("Image not found, check the file names and folder")
else:
    print("Image 1 size:", img1.shape)
    print("Image 2 size:", img2.shape)
```

## 3. Resizing
addWeighted needs both images to be of the same size, so image 2 is resized to match image 1.

```python
# cv2.resize takes (width, height), shape gives (height, width, channels)
height, width = img1.shape[:2]
img2 = cv2.resize(img2, (width, height))

print("Image 2 size after resize:", img2.shape)
```

## 4. Blending the images
Formula used: blended = 0.6 x image1 + 0.4 x image2 + 0

```python
blended = cv2.addWeighted(img1, 0.6, img2, 0.4, 0)
```

## 5. Saving the images

```python
cv2.imwrite("image1_resized.jpg", img1)
cv2.imwrite("image2_resized.jpg", img2)
cv2.imwrite("blended.jpg", blended)
```

## 6. Displaying the original and blended images

```python
# OpenCV reads images in BGR but matplotlib shows RGB, so converting before display
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.title("Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.title("Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
plt.title("Blended Image")
plt.axis("off")

plt.show()
```

## Output notes
- Image 2 was resized to the size of image 1 before blending
- Weights used are 0.6 for image 1 and 0.4 for image 2, so image 1 is slightly more visible
- Both input images and the blended image are saved in the folder (image1_resized.jpg, image2_resized.jpg, blended.jpg)
