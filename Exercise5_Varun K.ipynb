# Exercise 5 - Chatbot using Gradio

## 1. Installing and importing Gradio

```python

import gradio as gr
```

## 2. Chatbot function
This is a simple rule-based chatbot, so no API key is needed. It checks the words in the user's message and gives a matching reply.
gr.ChatInterface sends two things to this function: the new message and the chat history.

```python
def chatbot(message, history):
    msg = message.lower()

    if any(word in msg.split() for word in ["hi", "hello", "hey", "namaste"]):
        return "Hello! I am a simple chatbot. You can ask me about AI, machine learning, Gradio, Pandas or OpenCV."

    elif "artificial intelligence" in msg or "ai" in msg.split():
        return "Artificial Intelligence (AI) is the ability of computers to do tasks that normally need human intelligence, like understanding language, recognising images and making decisions."

    elif "machine learning" in msg:
        return "Machine Learning is a part of AI where a computer learns patterns from data instead of being given fixed rules."

    elif "gradio" in msg:
        return "Gradio is a Python library used to build simple web interfaces for models and functions, like this chatbot."

    elif "pandas" in msg:
        return "Pandas is a Python library for working with tables of data (DataFrames)."

    elif "opencv" in msg:
        return "OpenCV is a library for image and video processing, used for tasks like resizing and blending images."

    elif "nic" in msg.split():
        return "NIC (National Informatics Centre) provides IT and e-governance support to the Government of India."

    elif any(word in msg for word in ["bye", "thanks", "thank you"]):
        return "Thank you! Have a nice day."

    else:
        return "Sorry, I did not understand that. Try asking about AI, machine learning, Gradio, Pandas or OpenCV." 
```

## 3. Creating the chat interface
Title, description and sample prompts are set here.

```python
demo = gr.ChatInterface(
    fn=chatbot,
    title="Simple AI Chatbot",
    description="Type a message and press Enter. You can ask about AI, machine learning, Gradio, Pandas or OpenCV, or click one of the sample prompts below.",
    examples=[
        "What is Artificial Intelligence?",
        "What is Machine Learning?",
        "Tell me about Gradio",
    ],
)
```

## 4. Launching the interface

```python
# opens the chatbot in the browser
demo.launch()
```
